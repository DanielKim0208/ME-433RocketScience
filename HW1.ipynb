##The Atmosphere function is provided by the Public Domain Atmosphere Software(PDAS) (https://www.pdas.com/atmos.html)

```python
import math
import numpy as np

def Atmosphere(alt):
    """ Compute temperature, density, and pressure in standard atmosphere.
    Correct to 86 km.  Only approximate thereafter.
    Input:
        alt     geometric altitude, km.
    Return: (sigma, delta, theta)
        sigma   density/sea-level standard density
        delta   pressure/sea-level standard pressure
        theta   temperature/sea-level std. temperature
    """

    REARTH = 6369.0
    GMR = 34.163195
    NTAB = 8

    htab = [0.0, 11.0, 20.0, 32.0, 47.0,
            51.0, 71.0, 84.852]
    ttab = [288.15, 216.65, 216.65, 228.65, 270.65,
            270.65, 214.65, 186.946]
    ptab = [1.0, 2.2336110E-1, 5.4032950E-2, 8.5666784E-3,
            1.0945601E-3, 6.6063531E-4, 3.9046834E-5, 3.68501E-6]
    gtab = [-6.5, 0.0, 1.0, 2.8, 0, -2.8, -2.0, 0.0]

    h = alt * REARTH / (alt + REARTH)

    i = 0
    j = len(htab)

    while (j > i + 1):
        k = (i + j) // 2       # FIX: integer division
        if h < htab[k]:
            j = k
        else:
            i = k

    tgrad = gtab[i]
    tbase = ttab[i]
    deltah = h - htab[i]
    tlocal = tbase + tgrad * deltah
    theta = tlocal / ttab[0]

    if 0.0 == tgrad:
        delta = ptab[i] * math.exp(-GMR * deltah / tbase)
    else:
        delta = ptab[i] * math.pow(tbase / tlocal, GMR / tgrad)

    sigma = delta / theta

    return (sigma, delta, theta)
```

### Q2 - Kikiki altitude

```python
x = 0.00

for i in np.arange(0, 100, 0.01):
    sigma, delta, theta = Atmosphere(i)
    pressure = delta * 101325
    temperature = theta * 288.15

    x = temperature / pressure

    if x >= 71.547:
        break

print(i)
```

### Q4 - Hot Air Ballon

```python
x = 0.00

for i in np.arange(0, 50, 0.01):
    sigma, delta, theta = Atmosphere(i)
    pressure = delta * 101325
    temperature = theta * 288.15

    x = pressure * (1 / temperature )

    if x <= 21.060:
        break

print(i)
```
